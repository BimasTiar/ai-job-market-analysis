## Section 2 Skill Demand Analysis

```python
# ============================================================
# Question  : What are the most in-demand AI skills globally?
# Variables : skill, skill_category, skill_level
# Method    : Frequency aggregation + category breakdown
# ============================================================

# 2.1 Skill frequency
skill_freq = (df_skills_clean['skill']
              .value_counts()
              .reset_index())
skill_freq.columns = ['skill', 'count']
skill_freq['percentage'] = (skill_freq['count'] / len(df_skills_clean) * 100).round(2)

# 2.2 Category breakdown
cat_freq = (df_skills_clean['skill_category']
            .value_counts()
            .reset_index())
cat_freq.columns = ['skill_category', 'count']
cat_freq['percentage'] = (cat_freq['count'] / len(df_skills_clean) * 100).round(2)

# 2.3 Skill level by category
level_cat = (df_skills_clean
             .groupby(['skill_category','skill_level'])
             .size()
             .reset_index(name='count'))

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle('Global AI Skill Demand Analysis', fontsize=14, fontweight='bold')

# Bar Chart skill frequency
colors = ['#1f4e79' if i < 3 else 'steelblue' for i in range(len(skill_freq))]
bars = axes[0].barh(skill_freq['skill'], skill_freq['count'],
                    color=colors, edgecolor='white')
axes[0].set_title('Skill Demand Frequency')
axes[0].set_xlabel('Job Postings Count')
axes[0].invert_yaxis()
for bar, pct in zip(bars, skill_freq['percentage']):
    axes[0].text(bar.get_width() + 200,
                 bar.get_y() + bar.get_height()/2,
                 f'{pct}%', va='center', fontsize=9)

# Pie Chart by category
axes[1].pie(cat_freq['count'], labels=cat_freq['skill_category'],
            autopct='%1.1f%%',
            colors=['#1f77b4','#ff7f0e','#2ca02c'],
            startangle=90)
axes[1].set_title('Skill Demand by Category')

plt.tight_layout()
plt.show()

# Insight
print("\nSKILL DEMAND INSIGHT")
print("=" * 50)
print(f"Top 3 skills: {list(skill_freq.head(3)['skill'])}")
ml_pct = df_skills_clean[df_skills_clean['skill_category']=='ML'].shape[0] / len(df_skills_clean) * 100
print(f"ML skills share: {ml_pct:.1f}% of all requirements")
```

## Section 3 Salary Trend Analysis

```python
# ============================================================
# Question  : How have AI job salaries trended globally 2020–2026?
# Variables : posted_year, salary_avg_usd, country
# Method    : YoY trend + country comparison
# ============================================================

# 3.1 Global trend
salary_by_year = (df_jobs_merged
                  .groupby('posted_year')['salary_avg_usd']
                  .agg(['mean','median','std'])
                  .reset_index())
salary_by_year.columns = ['year','mean','median','std']
salary_by_year['yoy_growth_%'] = salary_by_year['mean'].pct_change().mul(100).round(2)

# 3.2 By country
salary_country_year = (df_jobs_merged
                       .groupby(['posted_year','country'])['salary_avg_usd']
                       .mean()
                       .round(0)
                       .reset_index())
salary_country_year.columns = ['year','country','avg_salary']

fig, axes = plt.subplots(1, 2, figsize=(16, 5))
fig.suptitle('Global AI Salary Trends (2020–2026)', fontsize=14, fontweight='bold')

# Line Chart global mean & median
axes[0].plot(salary_by_year['year'], salary_by_year['mean'],
             marker='o', label='Mean', color='steelblue', linewidth=2)
axes[0].plot(salary_by_year['year'], salary_by_year['median'],
             marker='s', label='Median', color='orange', linewidth=2)
axes[0].fill_between(salary_by_year['year'],
                     salary_by_year['mean'] - salary_by_year['std'],
                     salary_by_year['mean'] + salary_by_year['std'],
                     alpha=0.1, color='steelblue')
axes[0].axvline(2024.5, color='red', linestyle='--', alpha=0.5, label='Projected →')
axes[0].yaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f'${x:,.0f}'))
axes[0].set_title('Global Mean & Median Salary by Year')
axes[0].set_xlabel('Year')
axes[0].set_ylabel('Salary (USD)')
axes[0].legend()

# Line Chart by country
for country in salary_country_year['country'].unique():
    d = salary_country_year[salary_country_year['country']==country]
    axes[1].plot(d['year'], d['avg_salary'], marker='o', label=country, linewidth=2)
axes[1].axvline(2024.5, color='red', linestyle='--', alpha=0.5, label='Projected →')
axes[1].yaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f'${x:,.0f}'))
axes[1].set_title('Avg Salary by Country (2020–2026)')
axes[1].set_xlabel('Year')
axes[1].set_ylabel('Salary (USD)')
axes[1].legend(fontsize=8)

plt.tight_layout()
plt.show()

print("\nYEAR-OVER-YEAR SALARY GROWTH")
print("=" * 50)
print(salary_by_year[['year','mean','yoy_growth_%']].to_string(index=False))
```

## Section 4 H1: Seniority Pay Jumps

```python
# ============================================================
# Hypothesis: Senior-level AI salaries grow faster than
#             Entry/Mid between 2020–2026
# Variables : experience_level, posted_year, salary_avg_usd
# Method    : Grouped trend + 2020 vs 2026 growth comparison
# ============================================================

salary_exp_year = (df_jobs_merged
                   .groupby(['posted_year','experience_level'])['salary_avg_usd']
                   .mean()
                   .round(0)
                   .reset_index())

# Growth 2020 vs 2026
gap = salary_exp_year[salary_exp_year['posted_year'].isin([2020,2026])]
gap_pivot = (gap.pivot(index='experience_level',
                       columns='posted_year',
                       values='salary_avg_usd'))
gap_pivot['growth_%'] = ((gap_pivot[2026] - gap_pivot[2020]) / gap_pivot[2020] * 100).round(2)

fig, axes = plt.subplots(1, 2, figsize=(16, 5))
fig.suptitle('H1: Seniority Pay Jumps (2020–2026)', fontsize=14, fontweight='bold')

colors_exp = {'Entry':'#2ca02c','Mid':'#ff7f0e','Senior':'#1f77b4'}
for level in ['Entry','Mid','Senior']:
    d = salary_exp_year[salary_exp_year['experience_level']==level]
    axes[0].plot(d['posted_year'], d['salary_avg_usd'],
                 marker='o', label=level,
                 color=colors_exp[level], linewidth=2.5)
axes[0].axvline(2024.5, color='red', linestyle='--', alpha=0.5, label='Projected →')
axes[0].yaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f'${x:,.0f}'))
axes[0].set_title('Salary by Experience Level Over Time')
axes[0].set_xlabel('Year')
axes[0].set_ylabel('Salary (USD)')
axes[0].legend()

x = range(len(gap_pivot))
w = 0.35
axes[1].bar([i-w/2 for i in x], gap_pivot[2020], w,
            label='2020', color='lightsteelblue', edgecolor='white')
axes[1].bar([i+w/2 for i in x], gap_pivot[2026], w,
            label='2026', color='steelblue', edgecolor='white')
axes[1].set_xticks(x)
axes[1].set_xticklabels(gap_pivot.index)
axes[1].yaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f'${x:,.0f}'))
axes[1].set_title('Salary: 2020 vs 2026 by Experience Level')
axes[1].set_ylabel('Salary (USD)')
axes[1].legend()
for i, (idx, row) in enumerate(gap_pivot.iterrows()):
    axes[1].text(i, row[2026] + 500, f"+{row['growth_%']}%",
                 ha='center', fontsize=10, color='steelblue', fontweight='bold')

plt.tight_layout()
plt.show()

# H1 Verdict
print("\nH1 TEST RESULT")
print("=" * 50)
print(gap_pivot[['growth_%']].to_string())
senior_g = gap_pivot.loc['Senior','growth_%']
entry_g  = gap_pivot.loc['Entry','growth_%']
if senior_g > entry_g:
    print(f"\n✅ H1 SUPPORTED: Senior grew {senior_g-entry_g:.1f}pp faster than Entry.")
else:
    print(f"\n❌ H1 NOT SUPPORTED")
```

## Section 5 H2: Remote Wage Premium

```python
# ============================================================
# Hypothesis: Remote roles command higher salary premiums
#             vs Hybrid and Onsite globally
# Variables : remote_type, salary_avg_usd, country
# Method    : Group comparison + boxplot + country breakdown
# ============================================================

PALETTE = {'Remote': '#1f4e79', 'Hybrid': '#2e75b6', 'Onsite': '#9dc3e6'}

remote_means = (df_jobs_merged
                .groupby('remote_type')['salary_avg_usd']
                .agg(['mean','median','std','count'])
                .round(0).reset_index())

remote_country = (df_jobs_merged
                  .groupby(['country','remote_type'])['salary_avg_usd']
                  .mean().round(0).reset_index())
remote_country.columns = ['country','remote_type','avg_salary']

fig, axes = plt.subplots(1, 2, figsize=(16, 5))
fig.suptitle('H2: Remote vs Onsite Salary Premium', fontsize=14, fontweight='bold')

# Boxplot
order = ['Remote', 'Hybrid', 'Onsite']
data_box = [df_jobs_merged[df_jobs_merged['remote_type']==r]['salary_avg_usd'] for r in order]
bp = axes[0].boxplot(data_box, labels=order, patch_artist=True,
                     medianprops=dict(color='red', linewidth=2))
for patch, r_type in zip(bp['boxes'], order):
    patch.set_facecolor(PALETTE[r_type])
    patch.set_alpha(0.85)
axes[0].yaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f'${x:,.0f}'))
axes[0].set_title('Salary Distribution by Remote Type')
axes[0].set_ylabel('Salary (USD)')

# Grouped bar by country
countries = remote_country['country'].unique()
x = np.arange(len(countries))
width = 0.25

for i, r_type in enumerate(order):
    vals = [remote_country[(remote_country['country']==c) &
                           (remote_country['remote_type']==r_type)]['avg_salary'].values
            for c in countries]
    vals = [v[0] if len(v) > 0 else 0 for v in vals]
    axes[1].bar(x + i*width, vals, width,
                label=r_type, color=PALETTE[r_type], edgecolor='white')

axes[1].set_xticks(x + width)
axes[1].set_xticklabels(countries, rotation=45)
axes[1].yaxis.set_major_formatter(plt.FuncFormatter(lambda x,_: f'${x:,.0f}'))
axes[1].set_title('Avg Salary by Remote Type per Country')
axes[1].set_xlabel('Country')
axes[1].set_ylabel('Salary (USD)')
axes[1].legend(title='Remote Type')

plt.tight_layout()
plt.show()

# H2 Verdict
print("\nH2 TEST RESULT")
print("=" * 50)
print(remote_means[['remote_type','mean','median']].to_string(index=False))
remote_val = remote_means.loc[remote_means['remote_type']=='Remote','mean'].values[0]
onsite_val = remote_means.loc[remote_means['remote_type']=='Onsite','mean'].values[0]
premium    = remote_val - onsite_val
top        = remote_means.loc[remote_means['mean'].idxmax(), 'remote_type']
print(f"\nRemote premium over Onsite: ${premium:,.0f}")
if top == 'Remote':
    print(f"✅ H2 SUPPORTED: Remote roles lead at ${remote_val:,.0f} avg.")
else:
    print(f"❌ H2 NOT SUPPORTED: {top} leads instead.")
```

## Section 7 K-Means Clustering

```python
features = ['salary_avg_usd', 'min_experience_years',
            'experience_level', 'remote_type',
            'role_category', 'company_size']

df_cluster = df_jobs_merged[features].copy()

le = LabelEncoder()
cat_cols = ['experience_level', 'remote_type', 'role_category', 'company_size']
for col in cat_cols:
    df_cluster[col] = le.fit_transform(df_cluster[col])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_cluster)
```

### Section 7.1 Elbow Method

```python
k_range = range(2, 11)
inertia = []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_scaled)
    inertia.append(km.inertia_)

plt.figure(figsize=(9, 5))
plt.plot(k_range, inertia, marker='o', color='steelblue', linewidth=2.5)
plt.title('Elbow Method Optimal Number of Clusters',
          fontsize=13, fontweight='bold')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.xticks(k_range)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Section 7.2 Silhoutte Analysis

```python
sil_scores = []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    score  = silhouette_score(
        X_scaled, labels,
        sample_size=10000,
        random_state=42
    )
    sil_scores.append(score)
    print(f"  K={k}: Silhouette = {score:.4f}")

plt.figure(figsize=(9, 5))
plt.plot(k_range, sil_scores, marker='s', color='orange', linewidth=2.5)
plt.title('Silhouette Score by K — Higher is Better',
          fontsize=13, fontweight='bold')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.xticks(k_range)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Section 7.3 Final Model

Adjust FINAL_K if elbow and silhouette suggest different K

```python
FINAL_K = 3
km_final = KMeans(n_clusters=FINAL_K, random_state=42, n_init=10)
df_jobs_merged['cluster'] = km_final.fit_predict(X_scaled)
```

### Section 7.4 PCA Visualization

```python
from sklearn.decomposition import PCA
import matplotlib.cm as cm
import matplotlib.patches as mpatches

pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

CLUSTER_COLORS = ['#1f4e79', '#c55a11', '#538135']
CLUSTER_LABELS = ['Elite Strategists', 'Rising Engineers', 'Digital Nomad Analysts']

fig, ax = plt.subplots(figsize=(11, 7))

for cluster_id in range(FINAL_K):
    mask = df_jobs_merged['cluster'] == cluster_id
    ax.scatter(X_pca[mask, 0], X_pca[mask, 1],
               color=CLUSTER_COLORS[cluster_id],
               label=f'Cluster {cluster_id}: {CLUSTER_LABELS[cluster_id]}',
               alpha=0.35, s=20, edgecolors='none')

# Plot cluster centers in PCA space
centers_pca = pca.transform(km_final.cluster_centers_)
for i, (cx, cy) in enumerate(centers_pca):
    ax.scatter(cx, cy,
               color=CLUSTER_COLORS[i],
               s=250, marker='*',
               edgecolors='white', linewidths=0.8, zorder=5)
    ax.annotate(f'C{i}', (cx, cy),
                textcoords='offset points', xytext=(8, 5),
                fontsize=10, fontweight='bold',
                color=CLUSTER_COLORS[i])

ax.set_title(f'K-Means Job Market Segmentation via PCA (K={FINAL_K})',
             fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel(f'PCA Component 1 ({pca.explained_variance_ratio_[0]*100:.1f}% variance)',
              fontsize=11)
ax.set_ylabel(f'PCA Component 2 ({pca.explained_variance_ratio_[1]*100:.1f}% variance)',
              fontsize=11)

legend = ax.legend(title='Segments', markerscale=2,
                   fontsize=9, title_fontsize=10,
                   framealpha=0.9, loc='upper right')
ax.grid(alpha=0.15)
ax.set_facecolor('#f9f9f9')
fig.patch.set_facecolor('white')

plt.tight_layout()
plt.show()

print(f"Total variance captured: {pca.explained_variance_ratio_.sum()*100:.1f}%")
```
